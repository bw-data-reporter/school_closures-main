### We can't seem to figure out which schools have *actually* closed

**A few possibilities:**
 - Find places with the same mailing address
 - Find places with the same email attached
 - Find places with the same phone number
 - Find places with the same/similar names in the same city/district

```python
import pandas as pd

#load schools data set
df = pd.read_csv('data/all_schools.csv')
df.head()
```

**How often does the same email address show up?**

```python
emails = df['EntityEmail'].value_counts().reset_index()
emails[emails['count'] > 1]
```

423 instances where the same email address was used for multiple schools -- let's explore some of them

```python
df[df['EntityEmail'] == 'jvanderbrook@lakecitytrojans.org']
```

It looks like almost all of them are actually different schools -- so email doesn't work

**Phone number** - it seems more likely that different schools would have different phone numbers

```python
pd.set_option('display.float_format', lambda x: '%.4f' % x)
phones = df['EntityPhone'].value_counts().reset_index()
phones[phones['count'] > 1]
```

```python
df[df['EntityPhone'] == 3136562600]
```

In that example, a bunch of charter schools in the same network use the same phone number.

```python
df[df['EntityPhone'] == 2486585600]
```

### For public schools, we can match with NCES data

```python
#load NCES data
nces = pd.read_csv('data/nces_schools_data.csv')
nces.head()
```

```python
#filter schools to public schools (this does include charter)
public = df[df['EntityTypeName'].isin(['LEA School', 'PSA School', 'ISD School', 'State School'])]
len(public)
```

```python
len(public['EntityMailingStreet'].unique())
```

```python
#filter to schools marked as closed
public_closed = public[public['ClosedSince2000'] == True]
len(public_closed)
```

```python
len(public_closed['EntityMailingStreet'].unique())
```

```python
#check formatting
nces['School Name']
```

```python
public_closed['EntityOfficialName']
```

```python
#find matches (on school name and district)
matches = pd.merge(public_closed, nces, left_on=['EntityOfficialName', 'DistrictOfficialName'], right_on=['School Name', 'District'], how='inner')
matches.head()
```

```python
matches['Status'].value_counts()
```

```python
len(matches)
```

**That's 56 public or charter schools marked as closed that match schools/districts in the NCES data marked as active**
